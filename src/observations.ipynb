At first we normalized the entire pose each frame in accordance to the shoulder length. That poses several problems 
1- if the signer tilts their posture ever so slightly all of the distances will change, why? because in respect to the camera the shoulder length would be smaller 
2- we are calculating the shoulder length each frame and normalizing each frame, that means every pose extracted from each frame is not standardised to any measurement 
That's why we'll reshift the landmarks to use only the length  first frame 

I had a couple of bugs aswell, I was normalizing twice (first in visualizer.py and then in shift_landmarks.py) after fixing those bugs and normalizing using the first frame everything seems to work
